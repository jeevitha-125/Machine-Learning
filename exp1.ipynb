```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn import datasets, tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report,ConfusionMatrixDisplay
```

```python
iris=datasets.load_iris(as_frame=True)
iris_df=iris.frame
iris_df['species']=iris_df['target'].map(dict(enumerate(iris.target_names)))
iris_df=iris_df.drop(columns=['target'])
print(iris_df.head())
```

```python
print(iris_df.isnull().any())
print(iris_df.describe())
```

```python
sns.pairplot(iris_df,hue='species')
plt.savefig("ex1_pairplot.png")
```

```python
x=iris.data.values
y=iris_df['species'].values
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.9,random_state=1)
```

```python
clf=tree.DecisionTreeClassifier(random_state=1)
clf.fit(x_train,y_train)
```

```python
prediction=clf.predict(x_test)
print(prediction)
```

```python
print(classification_report(y_test,prediction))
```

```python
fig,ax=plt.subplots(figsize=(6,5))
ConfusionMatrixDisplay.from_estimator(clf,x_test,y_test,ax=ax,colorbar=True)
plt.tight_layout()
plt.savefig("ex1_confmatrix.png")
```

```python
fig,ax=plt.subplots(figsize=(12,8))
tree.plot_tree(clf,feature_names=iris.feature_names,class_names=iris.target_names,filled=True,rounded=True,ax=ax)
plt.tight_layout()
plt.savefig("ex1_true.png")
```
